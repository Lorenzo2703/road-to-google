## 📚 Theory

Understand fundamental sorting algorithms:
- Merge Sort: O(N log N) time, O(N) space. Stable.
- Quick Sort: O(N log N) average time, O(N^2) worst case. O(log N) space.
- Counting Sort / Bucket Sort: O(N) time for restricted domain of inputs.
